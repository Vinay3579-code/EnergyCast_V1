# EnergyCast V1 — Step 13: Model Explainability & SHAP (Shapley Additive Explanations)

This notebook conducts in-depth feature attribution, game-theoretic interpretability, and physical market validation for the **Champion LightGBM Regressor** using exact **TreeSHAP** (Lundberg et al., Nature Machine Intelligence 2020).

### Core Analytical Goals:
1. **Exact TreeSHAP Computation**: Validate the efficiency and exact additivity property ($f(x) = E[f(x)] + \sum \phi_i$) across all 4,200 test hours.
2. **Global Feature Attribution**: Rank features by mean absolute SHAP value (₹/MWh) and ascertain directionality.
3. **Local Waterfall Case Studies**: Unpack feature attributions for 3 distinct real-world market regimes:
   - *Case A: Midday Solar Glut* (Depressed MCP)
   - *Case B: Evening Peak Ramp* (Rapid upward price movement)
   - *Case C: Severe System Scarcity* (MCP reaching regulatory cap)
4. **Merit-Order Non-Linearity**: Plot the empirical merit-order stack of Net Demand (MW) vs. SHAP impact, discovering the exponential hockey-stick inflection beyond ~160 GW.
5. **Domain Attribution**: Break down feature contributions across Physical Grid, Autoregressive Lags, Market Bidding, Calendar, and Weather.
